```r
# El archivo cibmeteo.txt se exporta a formato CSV estandarizado
 ruta_txt <- if (file.exists("cibmeteo.txt")) {
  "cibmeteo.txt"
} else if (file.exists("datos/cibmeteo.txt")) {
  "datos/cibmeteo.txt"
} else {
  "../datos/cibmeteo.txt"
}

cat("--- PASO 1: LEYENDO ARCHIVO DE TEXTO Y CONVIRTIENDO A CSV ---\n")
datos_txt <- read.delim(ruta_txt, header = TRUE, sep = "\t", stringsAsFactors = FALSE)

# Crear directorio datos/ si no existe y guardar cibmeteo.csv
if (!dir.exists("datos")) dir.create("datos")
ruta_csv <- "datos/cibmeteo.csv"
write.csv(datos_txt, file = ruta_csv, row.names = FALSE)
cat("¡Conversión exitosa! Archivo guardado en:", ruta_csv, "\n\n")
```

```r
## Cargando el conjunto de datos en csv
datos_meteo <- read.csv(ruta_csv, stringsAsFactors = FALSE)
```

```r
## Mostrando las primeras 10 filas del dataset
cat("--- PRIMERAS 10 FILAS DEL ARCHIVO CIBMETEO.CSV ---\n")
print(head(datos_meteo, 10))
```

```r
## Exploración de datos
cat("\n--- Dimensiones del dataset (filas x columnas) ---\n")
print(dim(datos_meteo))

cat("\n--- Estructura internal de las variables (str) ---\n")
str(datos_meteo)

cat("\n--- Resumen estadístico (summary) ---\n")
summary(datos_meteo)
```
